# SectionB – Filtering & Conditional Selection

# 1. Select all students who completed the test preparation course.

```python
# To show me all the rows where the 'test preparation course' is equal to 'completed'."
#selects only the "test preparation course" column.
# == "completed"  This compares every value in the column with "completed".
# Pandas returns True or False for each row.
# .loc tells Pandas: "Return only the rows where the condition is True."
students_df.loc[students_df["test preparation course"] == "completed"]
```

### 2. Find al students who scored above 70 in math.

```python
students_df.loc[students_df["math score"] > 70]
```

### 3. Retrieve female  students who scored at least 65 in all three subjects (use the & operator)

```python
female_students = students_df[
    (students_df["gender"] == "female") &
    (students_df["math score"] >= 65) &
    (students_df["reading score"] >= 65) &
    (students_df["writing score"] >= 65)
]
female_students
```

```python
### 4. Count how many students have free/reduced lunch
```

```python
free_lunch_count = students_df[students_df["lunch"] == "free/reduced"].shape[0]

print("Number of students with free/reduced lunch:", free_lunch_count)
```

## Section C – Group By & Aggregation

### 1. Compute average math score by gender.

```python
# Compute the average math score by gender
average_math_score = students_df.groupby("gender")["math score"].mean()
average_math_score
```

### 2. Compute average scores(math,reading,writing) by parental level of education.

```python
# Compute average math, reading, and writing scores
average_scores = students_df.groupby("parental level of education")[
    ["math score", "reading score", "writing score"]
].mean()

average_scores
```

### 3. Compare average scores of students with vs without test preparation(note:'none'refers to students who did not complete the test preparation course

```python
# Compare average scores by test preparation course
test_prep_scores = students_df.groupby("test preparation course")[
    ["math score", "reading score", "writing score"]
].mean()

print(test_prep_scores)
```

## Section D – Visualization

### 1. Plot histograms for : ● math score ● reading score ● writing score

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(students_df["math score"], bins=10)
plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(students_df["reading score"], bins=10)
plt.title("Distribution of reading score")
plt.xlabel("reading score")
plt.ylabel("Number of Students")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(students_df["writing score"], bins=10)
plt.title("Distribution of writing score")
plt.xlabel("writing score")
plt.ylabel("Number of Students")
plt.show()
```

### Bar Plot of Average Score by Gender

```python
avg_math_gender = students_df.groupby("gender")["math score"].mean()
```

```python
plt.figure(figsize=(6, 5))
plt.bar(avg_math_gender.index, avg_math_gender.values)
plt.title("Average Math Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Math Score")
plt.show()
```
